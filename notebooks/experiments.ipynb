# RAG Doctor experiments
Compare retrieval methods and inspect failure modes.

```python
from src.pipeline import RAGDoctor
doctor = RAGDoctor.from_directory('data/documents')
import json
eval_set = json.load(open('data/evaluation.json'))
rows = doctor.evaluate(eval_set, top_k=5, rerank=False)
rows[:3]
```

```python
import pandas as pd
df = pd.DataFrame(rows)
df.groupby('method')[['recall@k','mrr','nDCG']].mean().round(3)
```
